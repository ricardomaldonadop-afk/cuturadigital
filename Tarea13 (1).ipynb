Escribe un programa en Python que muestra al usario el menu siguiente:


1.   Saludar.
2.   Mostrar un mensaje.
3.   Salir

Mientras el usario no presione la opción **3**, el programa debe seguir mandando el menu. Si el usario presiona **1**, el programa pide el nombre al usario y muetra a pantalla un saludo con su nombre. Si el usario presiona **2**, el programa debe mandar a panta lla un mensaje personalizado por ustedes.

Cada una de las lienas de código escritas deben estar acompañadas de un comentario que explique su uso.



```python
#Escriebe tu código.
se inisializa en 0 para entrar al ciclo
opcion = 0
#ciclo while que repetirael menu mientras el usuario no eliga la opcion 3
while opcion != 3:
  #se crea una linea para ver el menu mas ordenado
  print("\n=====================")
  #otra linea para separar visual mente
  print("\n=====================")
  #muestra la opcion disponibles en le menu
  print("1. saludar")
  print("2. mostrar un mensaje")
  print("3. salir")
  #se pide al usuario que escriva su eleccion
  #el input regresa el texto por eso se convierte a numero con int()
  opcion = int(input("elige una opcion:"))
  #si el usuario eligue 1 sele pide su nombre y se le saluda
  if opcion == 1:
    nombre = input("¿cual es tu nombre?")
  #se solicita nombre al usuario
  print(f"hola, (nombre).!mucho gusto saludarte!")
  #si el usuario selecciona 2 se muestra un mensaje adicional
  elif opcion == 2
  print("este es un mensaje especial")
  #si el usuario selecciona la opcion 3 el programa termina
  elif opcion == 3
  print("saliendo del programa !hasta luego!")
  #si el usuario escribe un numero que no sea 1,2,3
  else:
    print("opcion no valida intenta nuevamente.")
```
